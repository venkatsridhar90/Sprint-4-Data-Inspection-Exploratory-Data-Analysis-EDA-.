# Question 18: Identify Identifier Columns


## Step 1: Understand the concept

Identifier columns contain values used to distinguish individual records.

Examples include customerID, EmployeeID, and TransactionID.

**Why is it important?**
- Helps identify individual records.
- Helps check identifier uniqueness.
- Helps avoid using meaningless ID values as predictive features without justification.

## Step 2: Demonstrate the concept

The telecom dataset contains a `customerID` column.

We will find column names containing `id` and count their unique values to understand whether they may be identifiers.


## Step 3: Implement the Concept

```python
# Identify potential identifier columns
identifier_columns = [
    col for col in df.columns
    if "id" in col.lower()
]

print("Potential Identifier Columns:")

for col in identifier_columns:
    print(
        col,
        "- Unique values:",
        df[col].nunique(),
        "out of",
        len(df),
        "rows"
    )
```

### Code Explanation

- `df.columns` retrieves all column names.
- `col.lower()` converts a column name to lowercase.
- `"id" in col.lower()` checks whether the name contains `id`.
- `df[col].nunique()` counts unique values.
- `len(df)` counts the dataset rows.

**Expected Insight:**

`customerID` should be identified as a potential identifier and typically has one unique value per customer.

**AI/ML Use:**

Identifiers are often excluded from model features if they do not provide meaningful predictive information. The decision depends on the dataset and business context.

# Question 19: Identify Potential Target Columns

## Step 1: Understand the concept

A target column is the value that a machine learning model is trained to predict.

Examples include customer churn, house prices, and disease status.

**Why is it important?**
- Defines the prediction goal.
- Helps determine the ML problem type.
- Helps choose appropriate evaluation metrics.

## Step 2: Demonstrate the concept

In the telecom customer dataset, `Churn` is the potential target column.

The goal is to predict whether a customer will leave the company based on customer information.


## Step 3: Implement the Concept

```python
# Identify and inspect the potential target column
potential_target_columns = ["Churn"]

for col in potential_target_columns:
    if col in df.columns:
        print("Potential Target Column:", col)
        print("Data Type:", df[col].dtype)
        print("Unique Values:", df[col].nunique())

        print("\nTarget Distribution:")
        print(df[col].value_counts())
    else:
        print(col, "was not found in the dataset.")
```

### Code Explanation

- `potential_target_columns` stores the target column name.
- `if col in df.columns` checks whether the column exists.
- `df[col].dtype` displays the data type.
- `df[col].nunique()` counts distinct target values.
- `df[col].value_counts()` counts the records in each target category.

**Expected Insight:**

The `Churn` column typically contains two categories: `Yes` and `No`.

**AI/ML Use:**

Predicting Churn is a binary classification task because the target contains two classes. Checking the target distribution helps us understand whether the classes are balanced.
