# Notebook 1 – Dataset Understanding

## Dataset Understanding?

Dataset understanding means studying the structure, content, and characteristics of a dataset before performing data analysis or building a machine learning model.

### Before working with data, we need to understand:

- What each row represents
- What each column represents
- Which columns are features
- Which column is the target
- Which features are numerical or categorical
- Which features are discrete or continuous
- Which columns are identifiers
- How many rows and columns are present
- Whether missing or incorrect values exist

Understanding these properties helps us choose the correct data cleaning, analysis, visualization, and machine learning techniques.

# Question 3 – What is a Target?

A target is the outcome or result that we want to predict or understand.

In the Telco Customer Churn dataset, `Churn` is the target because we want to predict whether a customer will leave the company.

## Step 2 – Demonstrate the Concept

# Question 9 – Continuous Features


## Step 1 – Understand the Concept

Continuous features represent measurements that can take decimal values within a range.

Examples include:
- Temperature
- Height
- Weight
- Income
- Charges

In the telecom dataset, `MonthlyCharges` and `TotalCharges` are continuous monetary measurements.

## Step 2 – Demonstrate the Concept

### Simple Example

A person's height can be:
- 160.5 cm
- 160.55 cm
- 160.555 cm

It can take many possible values.

### Business Example

A telecom customer's monthly charge can be:
- 50.25
- 70.50
- 99.95

### AI/ML Use Case

Continuous features can help models identify numerical patterns and relationships.

### Why are they Important?

They can be analyzed using:
- Mean
- Standard deviation
- Histograms
- Correlation
- Distribution analysis


## Step 3 – Implement the Concept

```python
df["TotalCharges"] = pd.to_numeric(df["TotalCharges"], errors="coerce")

continuous_features = df[["MonthlyCharges", "TotalCharges"]]

print(continuous_features.head())
print("\nSummary:")
print(continuous_features.describe())
```

### Code Explanation

`pd.to_numeric()` converts `TotalCharges` into a numerical data type.

`errors="coerce"` converts invalid or blank values into `NaN`.

`describe()` provides statistical information such as:
- Count
- Mean
- Standard deviation
- Minimum
- Maximum

### Insight

The summary helps us understand the distribution and range of customer charges.

### ML Impact

Properly converted numerical features can be used during model training and analysis.

## Step 2 – Demonstrate the Concept

### Simple Example

- Marks can describe the student's performance.

### Business Example

### AI/ML Use Case

Identifiers usually should not be given directly to the model because they normally do not contain useful predictive information.

### Why are they Important?

Identifying ID columns prevents us from accidentally using meaningless information during model training.


## Step 3 – Implement the Concept

```python
print(df["customerID"].head())

print("\nNumber of Unique Customer IDs:", df["customerID"].nunique())
print("Number of Records:", len(df))
```

### Code Explanation

`nunique()` counts the number of unique customer IDs.

If the number of unique IDs is close to the number of records, the column is likely an identifier.

### Insight

`customerID` is used to identify customers rather than describe customer behavior.

### ML Impact

The identifier should normally be removed before model training to avoid meaningless information.

# Question 13 – Dataset Dimensions


## Step 1 – Understand the Concept

Dataset dimensions describe the size of a dataset in terms of rows and columns.

- Rows represent records or observations.
- Columns represent features or variables.

For example, a dataset with 1,000 rows and 20 columns has 1,000 records and 20 variables.

## Step 2 – Demonstrate the Concept

### Simple Example

If a dataset has:

100 rows × 5 columns

It means:
- 100 records
- 5 variables

### Business Example

A telecom customer dataset contains many customer records and several customer-related attributes.

### AI/ML Use Case

Dataset dimensions help us understand how much data is available for training and testing a machine learning model.

### Why are Dimensions Important?

They help us:
- Understand dataset size.
- Estimate computational requirements.
- Decide whether more data is needed.
- Check whether the dataset has too many or too few features.


## Step 3 – Implement the Concept

```python
rows, columns = df.shape

print("Number of Rows:", rows)
print("Number of Columns:", columns)
print("Dataset Dimensions:", df.shape)
```

### Code Explanation

`df.shape` returns a tuple containing:

`(number of rows, number of columns)`

The first value represents the number of records.

The second value represents the number of columns.

### Insight

Knowing the dataset dimensions helps us understand the size and complexity of the data.

### ML Impact

Dataset size influences model training time, memory usage, and the amount of information available for learning.
