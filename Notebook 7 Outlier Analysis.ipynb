# Outlier Analysis

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("data.csv", encoding="cp1252")

print("Dataset shape:", df.shape)
display(df.head())

# Select important numerical features
numeric_features = ["Quantity", "UnitPrice"]

display(df[numeric_features].describe())
```

## What is an outlier?
An outlier is a data point that is unusually far from most other observations in a dataset.
### Example:
Imagine a retail dataset containing the following quantities:
2, 3, 4, 5, 6, 7, 8, 10, 12, 500

## Types of outliers
#### A. Univariate outliers
An unusual value in a single feature.

### Example:
Quantity = 500 when most quantities are much smaller.
#### B. Multivariate outliers
An observation that looks unusual when several features are considered together, even if its individual values are not extreme.

### Example:
A product's quantity and unit price may each appear reasonable separately, but their combination may be unusual for that product or transaction.
#### C. Global and contextual outliers
- Global outlier: Unusual compared with the dataset overall.
- Contextual outlier: Unusual only within a particular context, such as a product, country, or time period.

```python
# Select Quantity for univariate outlier detection
quantity = df["Quantity"].dropna()

Q1 = quantity.quantile(0.25)
Q3 = quantity.quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)

# Identify potential outliers
quantity_outliers = df[
    (df["Quantity"] < lower_bound) |
    (df["Quantity"] > upper_bound)
]

print("Potential outlier rows:", len(quantity_outliers))
print(quantity_outliers[["InvoiceNo", "StockCode", "Quantity", "UnitPrice"]].head(10))
```

## Detection technique 3 — Box plot
A box plot visually displays the distribution of a numerical variable and highlights observations beyond the usual whisker boundaries.

```python
plt.figure(figsize=(10, 4))

sns.boxplot(x=df["Quantity"])

plt.title("Box Plot of Quantity")
plt.xlabel("Quantity")
plt.tight_layout()
plt.show()
```

## Detection technique 4 — Scatter plot
A scatter plot helps investigate unusual observations across two numerical features.

For example, we can examine Quantity and UnitPrice together.

```python
plot_data = df[["Quantity", "UnitPrice"]].dropna()

# Sample observations for a clearer visualization
plot_data = plot_data.sample(
    n=min(3000, len(plot_data)),
    random_state=42
)

plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=plot_data,
    x="Quantity",
    y="UnitPrice",
    alpha=0.5
)

plt.title("Quantity vs UnitPrice — Potential Outliers")
plt.xlabel("Quantity")
plt.ylabel("UnitPrice")
plt.tight_layout()
plt.show()
```
