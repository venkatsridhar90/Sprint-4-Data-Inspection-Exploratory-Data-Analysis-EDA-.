# Distribution Analysis

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("data.csv", encoding="cp1252")

print("Dataset shape:", df.shape)
display(df.head())

# Select important numerical features
numeric_features = ["Quantity", "UnitPrice"]

display(df[numeric_features].describe())
```

## Normal distribution
A normal distribution is a symmetric, bell-shaped distribution. Values near the center are more frequent, while values far from the center are less frequent.

```python
# Create normally distributed example data
np.random.seed(42)

normal_data = np.random.normal(
    loc=50, scale=10, size=1000
)

sns.histplot(normal_data, kde=True, bins=30)
plt.title("Normal Distribution")
plt.xlabel("Value")
plt.show()
```

## Uniform distribution
A uniform distribution has values that occur with approximately equal frequency across a specified range in its ideal continuous form.

```python
np.random.seed(42)

uniform_data = np.random.uniform(
    low=0, high=100, size=1000
)

sns.histplot(uniform_data, bins=20)
plt.title("Uniform Distribution")
plt.xlabel("Value")
plt.show()
```

## Symmetric distribution
A distribution is symmetric when its left and right sides have approximately similar shapes around the center.
## Left-skewed distribution
A left-skewed distribution, also called a negatively skewed distribution, has a longer tail extending toward smaller values.
## Right-skewed distribution
A right-skewed distribution, also called a positively skewed distribution, has a longer tail extending toward larger values.

## Skewness
Skewness measures the asymmetry of a distribution.
Skewness	General interpretation
Near 0	Approximately symmetric
Positive	Right-skewed
Negative	Left-skewed




```python
for column in ["Quantity", "UnitPrice"]:
    values = df[column].dropna()

    print(f"\n{column}")
    print("Skewness:", round(values.skew(), 3))
```

## Kurtosis
Kurtosis describes aspects of a distribution's tails and its tendency to produce extreme observations relative to a normal distribution.

Pandas calculates excess kurtosis by default:
Excess kurtosis	General interpretation
Near 0	Similar kurtosis to a normal distribution
Positive	Heavier tails or greater propensity for extreme values
Negative	Lighter tails than a normal distribution

```python
for column in ["Quantity", "UnitPrice"]:
    values = df[column].dropna()

    print(f"\n{column}")
    print("Excess kurtosis:", round(values.kurtosis(), 3))
```
