# Question

### How can we study the relationship between two variables using Bivariate Analysis?

**In this notebook, we will analyze:**

1. Numerical vs Numerical
2. Numerical vs Categorical
3. Categorical vs Categorical

**Techniques used:**

- Scatter Plot
- Box Plot
- Violin Plot
- Grouped Bar Chart
- Cross Tabulation
  

# Step 1: Understand the concept

Bivariate Analysis means studying the relationship between two variables.

It helps us understand whether one variable changes when another variable changes.

### 1. Numerical vs Numerical

Two numerical variables are compared.

Example:
- Tenure vs Monthly Charges
- Tenure vs Total Charges

A Scatter Plot can be used to observe the relationship.

### 2. Numerical vs Categorical

One numerical variable and one categorical variable are compared.

Example:
- Monthly Charges vs Contract
- Total Charges vs Internet Service

Box Plot and Violin Plot can be used to compare distributions across categories.

### 3. Categorical vs Categorical

Two categorical variables are compared.

Example:
- Contract vs Churn
- Internet Service vs Churn

Cross Tabulation and Grouped Bar Chart can be used.

### Why is Bivariate Analysis important?

Bivariate Analysis helps identify relationships, differences, and patterns between variables.

In Machine Learning, understanding relationships between features and the target variable helps in feature selection and model building.

# Step 2: Demonstrate the concept

### Example 1: Numerical vs Numerical

Suppose we compare `tenure` and `TotalCharges`.

Customers with longer tenure may have higher total charges because they have used the service for a longer period.

A Scatter Plot can help us visually identify this relationship.

### Example 2: Numerical vs Categorical

Suppose we compare `MonthlyCharges` with `Contract`.

Customers with different contract types may have different monthly charge distributions.

A Box Plot or Violin Plot helps compare these distributions.

### Example 3: Categorical vs Categorical

Suppose we compare `Contract` and `Churn`.

We can check how many customers in each contract type stayed or left.

A Cross Tabulation and Grouped Bar Chart can help identify the pattern.

### Real-Time / AI-ML Use

In a customer churn prediction system:

- `tenure` and `TotalCharges` can help understand customer behavior.
- `Contract` and `MonthlyCharges` can help identify differences between customer groups.
- `Contract` and `Churn` can help identify customer segments with higher churn.

These relationships can be useful when selecting features for a machine learning model.

## Step 3: Implementation the concept

```python
# Import Required Libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load Dataset

df = pd.read_csv("WA_Fn-UseC_-Telco-Customer-Churn.csv")

print("Dataset Loaded Successfully!")
print("Dataset Shape:", df.shape)
```

# Numerical vs Numerical

## Scatter Plot

A Scatter Plot is used to study the relationship between two numerical variables.

Here, we will compare:

- Tenure
- Total Charges

```python
# Convert TotalCharges to numeric

df["TotalCharges"] = pd.to_numeric(
    df["TotalCharges"],
    errors="coerce"
)

# Create Scatter Plot

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="tenure",
    y="TotalCharges"
)

plt.title("Tenure vs Total Charges")
plt.xlabel("Tenure (Months)")
plt.ylabel("Total Charges")
plt.tight_layout()
plt.show()
```

# Code Explanation

- `sns.scatterplot()` creates a scatter plot.
- `x="tenure"` places tenure on the X-axis.
- `y="TotalCharges"` places total charges on the Y-axis.
- Each point represents one customer.

### Expected Insight

If the points generally move upward as tenure increases, it indicates a positive relationship between tenure and total charges.

This means customers who stay longer tend to accumulate higher total charges.

### AI/ML Use

This relationship can help a machine learning model understand customer spending behavior and identify useful features.

# Numerical vs Categorical

## Box Plot

A Box Plot compares the distribution of a numerical variable across different categories.

Here, we compare:

- Monthly Charges
- Contract

```python
# Create Box Plot

plt.figure(figsize=(9, 5))

sns.boxplot(
    data=df,
    x="Contract",
    y="MonthlyCharges"
)

plt.title("Monthly Charges by Contract Type")
plt.xlabel("Contract")
plt.ylabel("Monthly Charges")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

# Code Explanation

- `sns.boxplot()` creates a box plot.
- `Contract` is the categorical variable.
- `MonthlyCharges` is the numerical variable.
- The box shows the distribution of monthly charges for each contract type.
- The middle line represents the median.
- The points outside the whiskers may represent potential outliers.

### Expected Insight

The box plot helps us compare the typical monthly charges and variation across different contract types.

### AI/ML Use

This can help identify whether contract type is associated with different pricing patterns.

## Violin Plot

A Violin Plot shows the distribution and density of a numerical variable across categories.

We will compare Monthly Charges across Contract types.

```python
# Create Violin Plot

plt.figure(figsize=(9, 5))

sns.violinplot(
    data=df,
    x="Contract",
    y="MonthlyCharges"
)

plt.title("Monthly Charges Distribution by Contract Type")
plt.xlabel("Contract")
plt.ylabel("Monthly Charges")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

# Code Explanation

- `sns.violinplot()` shows the distribution of numerical values across categories.
- The width of the violin represents the density of observations.
- Wider areas indicate that more customers have values in that range.

### Expected Insight

The violin plot helps us understand where monthly charges are concentrated for each contract type.

### AI/ML Use

Distribution differences between customer groups can help identify useful patterns for customer segmentation and churn prediction.

# Categorical vs Categorical

## Cross Tabulation

Cross Tabulation is used to compare the frequency of two categorical variables.

Here, we will compare:

- Contract
- Churn

```python
# Create Cross Tabulation

cross_table = pd.crosstab(
    df["Contract"],
    df["Churn"]
)

print("Contract vs Churn:")
print(cross_table)
```

# Code Explanation

- `pd.crosstab()` creates a frequency table.
- Rows represent Contract types.
- Columns represent Churn status.
- The values represent the number of customers.

### Expected Insight

The table helps us compare how many customers stayed and how many customers churned within each contract type.

### AI/ML Use

This relationship can help identify customer groups that may have a higher risk of churn.

# Grouped Bar Chart

A Grouped Bar Chart visually compares two categorical variables.

Here, we compare Contract and Churn.

```python
# Create Grouped Bar Chart

cross_table.plot(
    kind="bar",
    figsize=(9, 5)
)

plt.title("Churn Distribution by Contract Type")
plt.xlabel("Contract")
plt.ylabel("Number of Customers")
plt.xticks(rotation=20)
plt.legend(title="Churn")
plt.tight_layout()
plt.show()
```

# Code Explanation

- `cross_table.plot(kind="bar")` creates a bar chart from the cross-tabulation.
- Each contract type contains separate bars for Churn = Yes and Churn = No.
- This makes it easier to compare customer churn across contract types.

### Expected Insight

The chart helps identify which contract types have more customers leaving or staying.

### AI/ML Use

This type of relationship analysis can help businesses identify customer segments that require retention strategies.

# Correlation Between Numerical Variables

Correlation measures the strength and direction of the relationship between two numerical variables.

We will calculate the correlation between:

- Tenure
- Monthly Charges
- Total Charges

```python
# Select Numerical Variables

numerical_data = df[
    ["tenure", "MonthlyCharges", "TotalCharges"]
].copy()

# Calculate Correlation

correlation = numerical_data.corr()

print("Correlation Matrix:")
print(correlation.round(2))
```

# Code Explanation

- `.corr()` calculates the correlation between numerical variables.
- Correlation values range from -1 to +1.

### Interpretation

- Close to +1 → Strong positive relationship
- Close to -1 → Strong negative relationship
- Close to 0 → Weak or no linear relationship

### AI/ML Use

Correlation analysis helps identify relationships between numerical features and can support feature selection before machine learning.
