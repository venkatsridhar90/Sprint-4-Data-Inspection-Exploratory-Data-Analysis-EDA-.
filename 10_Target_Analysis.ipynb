# Notebook 10 – Target Variable Analysis

**Dataset:** Telco Customer Churn Dataset

**Target Variable:** Churn

This notebook analyzes the target variable separately to understand its classes, distribution, percentages, and class imbalance before Machine Learning.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("WA_Fn-UseC_-Telco-Customer-Churn.csv")

print("Dataset Loaded Successfully!")
print("Dataset Shape:", df.shape)
```

## Question 1 – What is the Target Variable?

### Step 1: Understand the concept

A target variable is the variable that a Machine Learning model is trained to predict.

In the Telco Customer Churn dataset, the target variable is `Churn`.

- `Yes` → Customer churned
- `No` → Customer did not churn

This is a classification problem because the target contains categorical classes.

### Step 2: Demonstrate the concept

Customer information such as tenure, contract, monthly charges, and internet service can be used to predict whether a customer will churn.

### Step 3: Implementation the concept

We identify the target variable and check its unique values.

```python
target_variable = "Churn"

print("Target Variable:", target_variable)
print("Unique Target Values:", df[target_variable].unique())
print("Number of Classes:", df[target_variable].nunique())
```

### Code Explanation

The target variable is `Churn`.

It contains two classes: `Yes` and `No`.

Therefore, this is a binary classification problem.

## Question 2 – What is Class Distribution?

### Step 1: Understand the concept

Class distribution shows how the observations are divided among the different target classes.

For the `Churn` variable, we want to understand how many customers:

- Churned (`Yes`)
- Did not churn (`No`)

### Step 2: Demonstrate the concept

If a dataset contains:

- No → 80%
- Yes → 20%

then the target classes are not equally distributed.

### Step 3: Implementation the concept

We calculate the number of observations belonging to each class.

```python
class_distribution = df["Churn"].value_counts()

print("Class Distribution:")
print(class_distribution)
```

### Code Explanation

`value_counts()` counts the number of records in each target class.

This helps us understand the distribution of `Yes` and `No` customers.

## Question 3 – What are the Class Counts?

### Step 1: Understand the concept

Class count represents the total number of observations present in each target class.

For the Churn variable:

- `Yes` represents customers who churned.
- `No` represents customers who stayed.

### Step 2: Demonstrate the concept 

For example:

- No → 5000 customers
- Yes → 2000 customers

These values represent the class counts.

### Step 3: Implementation the concept

We calculate the count of each Churn class.

```python
class_counts = df["Churn"].value_counts()

print("Class Counts:")
print(class_counts)
```

### Code Explanation

The output shows the number of customers in each target class.

The larger count represents the majority class, while the smaller count represents the minority class.

## Question 4 – What are Class Percentages?

### Step 1: Understand the concept

Class percentage shows what percentage of the complete dataset belongs to each target class.

Formula:

Class Percentage = (Class Count / Total Records) × 100

### Step 2: Demonstrate the concept

For example, if 200 out of 1000 customers churn:

Class Percentage = (200 / 1000) × 100 = 20%

### Step 3: Implementation the concept

We calculate the percentage of each Churn class.

```python
class_percentages = df["Churn"].value_counts(normalize=True) * 100

print("Class Percentages:")
print(class_percentages.round(2))
```

### Code Explanation

`normalize=True` converts the class counts into proportions.

Multiplying by 100 converts the proportions into percentages.

## Question 5 – How can we Visualize Class Distribution?

## Question 6 – How can we Visualize Class Percentages?

## Question 7 – What is Class Imbalance?

### Step 1: Understand the concept

Class imbalance occurs when one target class contains significantly more observations than another target class.

For example:

- Class A → 90%
- Class B → 10%

This represents an imbalanced target.

### Step 2: Demonstrate the concept

In a customer churn problem, if most customers belong to the `No` class and fewer customers belong to the `Yes` class, the target is imbalanced.

### Step 3: Implementation the concept

We compare the percentages of the largest and smallest classes.

```python
class_percentages = df["Churn"].value_counts(normalize=True) * 100

largest_class_percentage = class_percentages.max()
smallest_class_percentage = class_percentages.min()

print("Largest Class Percentage:", round(largest_class_percentage, 2), "%")
print("Smallest Class Percentage:", round(smallest_class_percentage, 2), "%")
```

### Code Explanation

The difference between the largest and smallest class percentages helps us understand whether the target classes are balanced or imbalanced.

## Question 8 – How can we Calculate the Class Imbalance Ratio?

### Step 1: Understand the concept

The imbalance ratio compares the majority class count with the minority class count.

Formula:

Imbalance Ratio = Majority Class Count / Minority Class Count

### Step 2: Demonstrate the concept

For example:

Majority Class = 800  
Minority Class = 200

Imbalance Ratio = 800 / 200 = 4

This means the majority class has four times more observations.

### Step 3: Implementation the concept

We calculate the imbalance ratio for the Churn target.

```python
class_counts = df["Churn"].value_counts()

majority_count = class_counts.max()
minority_count = class_counts.min()

imbalance_ratio = majority_count / minority_count

print("Majority Class Count:", majority_count)
print("Minority Class Count:", minority_count)
print("Imbalance Ratio:", round(imbalance_ratio, 2))
```

### Code Explanation

The imbalance ratio shows how much larger the majority class is compared with the minority class.

##  Which are the Majority and Minority Classes?

### Step 1: Understand the concept

The **majority class** is the class with the highest number of observations.

The **minority class** is the class with the lowest number of observations.

### Step 2: Demonstrate the concept

For the Churn target:

- The class with more customers → Majority Class
- The class with fewer customers → Minority Class

### Step 3: Implementation the concept

We identify both classes using their counts.

```python
class_counts = df["Churn"].value_counts()

majority_class = class_counts.idxmax()
minority_class = class_counts.idxmin()

print("Majority Class:", majority_class)
print("Minority Class:", minority_class)
```

### Code Explanation

`idxmax()` returns the class with the highest count.

`idxmin()` returns the class with the lowest count.

## Question 9 – What is Target Distribution in Regression?

Target distribution shows how the target values are spread across different ranges.

It helps us understand:

- Center of the values

### Step 2: Demonstrate the concept

100000, 150000, 200000, 250000, 500000

A histogram can be used to understand how these values are distributed.

We use a histogram to visualize a regression target distribution.

### Code Explanation

## Question 10 – What is the Mean of a Regression Target?

### Step 1: Understand the concept

Mean represents the average value of the regression target.

Formula:

Mean = Sum of all values / Number of values

### Step 2: Demonstrate the concept

For example:

Values = 10, 20, 30

Mean = (10 + 20 + 30) / 3
     = 20

### Step 3: Implementation the concept

We calculate the mean of the regression target.

### Code Explanation

## Question 11 – What is the Median of a Regression Target?

### Step 1: Understand the concept

Median is the middle value when the target values are arranged in ascending order.

Median is useful when the target contains extreme values because it is less affected by outliers than the mean.

### Step 2: Demonstrate the concept

Example:

10, 20, 30, 40, 100

Median = 30

The extreme value 100 affects the mean more than the median.

### Step 3: Implementation the concept

We calculate the median of the regression target.

```python
median_value = regression_target.median()

print("Median:", round(median_value, 2))
```

## Question 12 – What is the Variance of a Regression Target?

### Step 1: Understand the concept

Variance measures how far the target values are spread from the mean.

A higher variance indicates greater variation in the target values.

### Step 2: Demonstrate the concept

If target values are very close to the mean, variance will be small.

If target values are widely spread, variance will be large.

### Step 3: Implementation the concept

We calculate the variance of the regression target.

```python
variance_value = regression_target.var()

print("Variance:", round(variance_value, 2))
```

### Code Explanation

Variance measures the overall spread of the regression target around its mean.

## Question 13 – What is the Standard Deviation of a Regression Target?

### Step 1: Understand the concept

Standard deviation measures how much the target values typically differ from the mean.

A higher standard deviation means the target values are more spread out.

### Step 2: Demonstrate the concept

Standard deviation is the square root of variance.

### Step 3: Implementation the concept

We calculate the standard deviation of the regression target.

```python
std_value = regression_target.std()

print("Standard Deviation:", round(std_value, 2))
```

### Code Explanation

Standard deviation gives the spread of the target in the same unit as the original target variable.

## Question 14 – What is Skewness of a Regression Target?

### Step 1: Understand the concept

Skewness measures the asymmetry of the target distribution.

- Skewness ≈ 0 → Approximately symmetric
- Positive skewness → Right-skewed
- Negative skewness → Left-skewed

### Step 2: Demonstrate the concept

A right-skewed target has a longer tail toward larger values.

A left-skewed target has a longer tail toward smaller values.

### Step 3: Implementation the concept

We calculate the skewness of the regression target.

```python
skewness_value = regression_target.skew()

print("Skewness:", round(skewness_value, 2))

if skewness_value > 0.5:
    print("Distribution: Right-Skewed")
elif skewness_value < -0.5:
    print("Distribution: Left-Skewed")
else:
    print("Distribution: Approximately Symmetric")
```

### Code Explanation

The skewness value helps us understand whether the regression target is symmetric or has a longer tail on either side.

## Question 15 – How can we Identify Outliers in a Regression Target?

### Step 1: Understand the concept 

An outlier is a target value that is unusually far from most other target values.

The IQR method can be used to identify potential outliers.

### Step 2: Demonstrate the concept

The IQR method uses:

- Q1 → 25th percentile
- Q3 → 75th percentile
- IQR = Q3 - Q1

Potential outliers are values below the lower bound or above the upper bound.

### Step 3: Implementation the concept

We identify potential outliers using the IQR method.

```python
Q1 = regression_target.quantile(0.25)
Q3 = regression_target.quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = regression_target[
    (regression_target < lower_bound) |
    (regression_target > upper_bound)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)

print("Number of Outliers:", len(outliers))
print("\nOutlier Values:")
print(outliers.values)
```

### Code Explanation

The IQR method identifies unusually high or low target values.

Outliers should not automatically be removed. They may represent genuine observations, unusual business cases, or data errors.
