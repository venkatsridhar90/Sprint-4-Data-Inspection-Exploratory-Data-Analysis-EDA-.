# Question 1: Calculate Mean, Median and Mode


## Step 1: Understand the concept

Mean is the average value of a variable.

Median is the middle value after sorting the data.

Mode is the most frequently occurring value.

These measures help us understand the central tendency of numerical data.

For example, in customer analysis, the mean MonthlyCharges can show the average monthly amount paid by customers.

## Step 2: Demonstrate the concept

For the Telco Customer Churn dataset:

- `tenure` shows how long a customer has stayed with the company.
- `MonthlyCharges` shows the customer's monthly charge.
- `TotalCharges` shows the total amount charged to the customer.

Comparing mean and median helps us understand whether the data is balanced or affected by extreme values.


## Step 3: Implement the Concept

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("WA_Fn-UseC_-Telco-Customer-Churn.csv")

print("Dataset Loaded Successfully!")
print("Dataset Shape:", df.shape)
```

```python
# Calculate Mean, Median and Mode

numerical_columns = ["tenure", "MonthlyCharges", "TotalCharges"]

for col in numerical_columns:
    data = pd.to_numeric(df[col], errors="coerce")

    print(f"\n--- {col} ---")
    print("Mean:", data.mean())
    print("Median:", data.median())
    print("Mode:", data.mode().iloc[0])
```

## Code Explanation

- `pd.to_numeric()` converts values into numeric form.
- `mean()` calculates the average.
- `median()` finds the middle value.
- `mode()` finds the most frequently occurring value.

### AI/ML Use

Central tendency helps us understand the typical behavior of customers and can be useful during feature analysis before building a machine learning model.

# Question 2: Calculate Minimum, Maximum and Range


## Step 1: Understand the concept

Minimum is the smallest value.

Maximum is the largest value.

Range shows the difference between the maximum and minimum values.

Range = Maximum - Minimum

Range helps us understand the overall spread of a variable.

## Step 2: Demonstrate the concept

For example, `MonthlyCharges` can show the lowest and highest monthly charges paid by customers.

A large range indicates that customers have very different charge levels.


## Step 3: Implement the Concept

```python
# Calculate Minimum, Maximum and Range

for col in numerical_columns:
    data = pd.to_numeric(df[col], errors="coerce")

    minimum = data.min()
    maximum = data.max()
    data_range = maximum - minimum

    print(f"\n--- {col} ---")
    print("Minimum:", minimum)
    print("Maximum:", maximum)
    print("Range:", data_range)
```

## Code Explanation

- `min()` finds the smallest value.
- `max()` finds the largest value.
- Maximum minus Minimum gives the range.

### AI/ML Use

Range helps identify the scale and spread of features before preprocessing or model training.

# Question 3: Calculate Variance and Standard Deviation


## Step 1: Understand the concept

Variance measures how far values are spread from the mean.

Standard deviation is the square root of variance.

A higher standard deviation means the values are more spread out.

A lower standard deviation means the values are closer to the mean.




## Step 2: Demonstrate the concept

For example, if MonthlyCharges has a high standard deviation, customer monthly charges vary considerably.

This information is useful when understanding customer behavior and preparing numerical features for machine learning.


## Step 3: Implement the Concept

```python
# Calculate Variance and Standard Deviation

for col in numerical_columns:
    data = pd.to_numeric(df[col], errors="coerce")

    print(f"\n--- {col} ---")
    print("Variance:", data.var())
    print("Standard Deviation:", data.std())
```

## Code Explanation

- `var()` calculates variance.
- `std()` calculates standard deviation.

### AI/ML Use

Variance and standard deviation help identify feature spread and are important when applying techniques such as feature scaling and normalization.

# Question 4: Calculate Quartiles


## Step 1: Understand the concept

Quartiles divide numerical data into four parts.

- Q1 = 25th percentile
- Q2 = 50th percentile (Median)
- Q3 = 75th percentile

The Interquartile Range (IQR) is:

IQR = Q3 - Q1

IQR represents the spread of the middle 50% of the data.

## Step 2: Demonstrate the concept

For customer data, quartiles can help us understand the distribution of customer tenure or monthly charges.

For example, Q1 of MonthlyCharges represents the point below which approximately 25% of customers fall.


## Step 3: Implement the Concept

```python
# Calculate Quartiles

for col in numerical_columns:
    data = pd.to_numeric(df[col], errors="coerce")

    q1 = data.quantile(0.25)
    q2 = data.quantile(0.50)
    q3 = data.quantile(0.75)

    print(f"\n--- {col} ---")
    print("Q1 (25%):", q1)
    print("Q2 (50%):", q2)
    print("Q3 (75%):", q3)
    print("IQR:", q3 - q1)
```

## Code Explanation

- `quantile(0.25)` calculates Q1.
- `quantile(0.50)` calculates Q2.
- `quantile(0.75)` calculates Q3.
- Q3 - Q1 gives the IQR.

### AI/ML Use

Quartiles and IQR are useful for understanding distributions and detecting outliers.

# Question 5: Calculate Percentiles


## Step 1: Understand the concept

A percentile tells us the value below which a certain percentage of observations fall.

For example:

- 10th percentile → 10% of values are below this point.
- 50th percentile → 50% of values are below this point.
- 90th percentile → 90% of values are below this point.

## Step 2: Demonstrate the concept

In customer analysis, the 90th percentile of MonthlyCharges can help identify customers with relatively high monthly charges.


## Step 3: Implement the Concept

```python
# Calculate Percentiles

percentiles = [0.10, 0.25, 0.50, 0.75, 0.90]

for col in numerical_columns:
    data = pd.to_numeric(df[col], errors="coerce")

    print(f"\n--- {col} ---")

    for p in percentiles:
        print(f"{int(p * 100)}th Percentile:", data.quantile(p))
```

## Code Explanation

`quantile()` calculates the requested percentile.

Percentiles help us understand where observations fall within the overall distribution.

### AI/ML Use

Percentiles can be used for feature analysis, outlier detection, and creating business thresholds.

# Question 6: Calculate Skewness and Kurtosis


## Step 1: Understand the concept

Skewness measures the asymmetry of a distribution.

- Skewness near 0 → approximately symmetric
- Positive skewness → longer right tail
- Negative skewness → longer left tail

Kurtosis describes the shape and tail behavior of a distribution.

These measures help us understand the distribution beyond mean and standard deviation.

## Step 2: Demonstrate the concept

If MonthlyCharges has positive skewness, some customers may have relatively high monthly charges.

Skewness and kurtosis can help identify unusual distribution patterns before machine learning.


## Step 3: Implement the Concept

```python
# Calculate Skewness and Kurtosis

for col in numerical_columns:
    data = pd.to_numeric(df[col], errors="coerce")

    print(f"\n--- {col} ---")
    print("Skewness:", data.skew())
    print("Kurtosis:", data.kurt())
```

## Code Explanation

- `skew()` calculates skewness.
- `kurt()` calculates kurtosis.

### AI/ML Use

Skewness can help decide whether a transformation may be useful for a numerical feature before modeling.

# Question 8: Create KDE / Density Plot


## Step 1: Understand the concept

A KDE (Kernel Density Estimate) plot is a smooth representation of the distribution of numerical data.

It helps us understand where values are concentrated.

## Step 2: Demonstrate the concept

A KDE plot of tenure can show where customer tenure values are most concentrated.

This gives a smoother view than a histogram.


## Step 3: Implement the Concept

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("WA_Fn-UseC_-Telco-Customer-Churn.csv")

print("Dataset Shape:", df.shape)
```

```python
# Define numerical columns

numerical_columns = [
    "tenure",
    "MonthlyCharges",
    "TotalCharges"
]

print("Numerical Columns:", numerical_columns)
```

```python
for col in numerical_columns:
    data = pd.to_numeric(df[col], errors="coerce").dropna()

    plt.figure(figsize=(8, 5))

    sns.kdeplot(
        x=data,
        fill=True
    )

    plt.title(f"Density Plot of {col}")
    plt.xlabel(col)
    plt.ylabel("Density")
    plt.show()
```

## Code Explanation

- `kdeplot()` creates a density plot.
- `fill=True` fills the area under the density curve.

### AI/ML Use

Density plots help identify concentration, skewness, and unusual distribution patterns in numerical features.

# Question 12: Create Bar Charts


## Step 1: Understand the concept

A bar chart compares the frequency of different categories.

It is useful for categorical variables because each category is represented by a separate bar.

## Step 2: Demonstrate the concept

A bar chart of Contract can quickly show which contract type has the highest number of customers.


## Step 3: Implement the Concept

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


# Create Bar Charts

for col in categorical_columns:
    plt.figure(figsize=(9, 5))

    value_counts = df[col].value_counts()

    value_counts.plot(kind="bar")

    plt.title(f"Frequency of {col}")
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```

## Code Explanation

- `value_counts()` calculates category frequencies.
- `plot(kind="bar")` creates a bar chart.
- `rotation=45` improves readability of long category names.

### AI/ML Use

Bar charts help identify dominant and rare categories before feature encoding.

# Question 14: Create a Pie Chart


## Step 1: Understand the concept

A pie chart represents the percentage contribution of each category to the whole.

It is appropriate when a categorical variable has a small number of meaningful categories.

## Step 2: Demonstrate the concept

Churn is suitable for a pie chart because it contains two main categories:

- Yes
- No

The chart can show the percentage of customers who churned and stayed.


## Step 3: Implement the Concept

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


# Create Pie Chart for Churn

churn_counts = df["Churn"].value_counts()

plt.figure(figsize=(7, 7))

plt.pie(
    churn_counts,
    labels=churn_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Customer Churn Distribution")
plt.show()
```

## Code Explanation

- `value_counts()` calculates the number of customers in each Churn category.
- `pie()` creates the pie chart.
- `autopct` displays percentages.
- `startangle=90` rotates the chart for better presentation.

### AI/ML Use

The Churn distribution helps us understand whether the target variable is balanced or imbalanced before building a classification model.

# Question 15: Explain Findings for Important Variables


## Step 1: Understand the concept

After calculating statistical measures and creating visualizations, we should interpret what the data tells us.

The goal of univariate analysis is not only to calculate values but also to understand the behavior of each variable.

## Step 2: Demonstrate the concept

Important numerical variables:

- `tenure` → customer relationship duration
- `MonthlyCharges` → monthly customer charges
- `TotalCharges` → total amount charged

Important categorical variables:

- `Contract` → customer contract type
- `InternetService` → internet service type
- `PaymentMethod` → payment method
- `Churn` → customer churn status


## Step 3: Implement the Concept

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns



# Generate a simple numerical summary for important variables

summary = pd.DataFrame({
    "Mean": df[numerical_columns].apply(pd.to_numeric, errors="coerce").mean(),
    "Median": df[numerical_columns].apply(pd.to_numeric, errors="coerce").median(),
    "Minimum": df[numerical_columns].apply(pd.to_numeric, errors="coerce").min(),
    "Maximum": df[numerical_columns].apply(pd.to_numeric, errors="coerce").max(),
    "Std_Dev": df[numerical_columns].apply(pd.to_numeric, errors="coerce").std(),
    "Skewness": df[numerical_columns].apply(pd.to_numeric, errors="coerce").skew()
})

print("Numerical Variable Summary:")
print(summary.round(2))

print("\nCategorical Variable Summary:")

for col in categorical_columns:
    print(f"\n{col}:")
    print(df[col].value_counts(normalize=True).mul(100).round(2))
```

- Standard deviation
- Skewness
